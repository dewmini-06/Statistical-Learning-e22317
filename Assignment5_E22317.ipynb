Part A: Theoretical Fundamentals (Maximum Likelihood & Decision Space)

## 1. Derivation of the Bias in the MLE Covariance Matrix

We begin with the Maximum Likelihood Estimator (MLE) for the population covariance matrix $\boldsymbol{\Sigma}$:

$$\widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$$

where $\widehat{\boldsymbol{\mu}}_n = \frac{1}{n}\sum_{i=1}^n \mathbf{X}_i$ is the empirical sample mean.

To determine if this estimator is biased, we expand the deviations around the true, unobservable population mean $\boldsymbol{\mu}$:

$$\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n = (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$$

Substituting this back into the sum inside the MLE formula gives:

$$\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T = \sum_{i=1}^n \left[ (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right] \left[ (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right]^T$$

Expanding the product matrix term-by-term yields:

$$= \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T - \sum_{i=1}^n (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T + \sum_{i=1}^n (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

Notice that $\sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu}) = n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$. Because $(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$ does not depend on the index $i$, we can simplify the cross-product terms:

$$\sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T = \left[ \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu}) \right] (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T = n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

Substituting this simplification back into the expanded expression:

$$= \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T - n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T + n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

$$= \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

Now, we divide by $n$ to return to $\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}$ and apply the expectation operator $\mathbb{E}[\cdot]$ to both sides:

$$\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{1}{n} \sum_{i=1}^n \mathbb{E}[(\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T] - \mathbb{E}[(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T]$$

By definition of the structural population moments for i.i.d. variables:
* $\mathbb{E}[(\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T] = \operatorname{Var}(\mathbf{X}_i) = \boldsymbol{\Sigma}$
* $\mathbb{E}[(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T] = \operatorname{Var}(\widehat{\boldsymbol{\mu}}_n) = \frac{1}{n}\boldsymbol{\Sigma}$

Substituting these distributions back into our expectation equation gives:

$$\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{1}{n} (n\boldsymbol{\Sigma}) - \frac{1}{n}\boldsymbol{\Sigma} = \boldsymbol{\Sigma} - \frac{1}{n}\boldsymbol{\Sigma} = \left(\frac{n-1}{n}\right)\boldsymbol{\Sigma}$$

Because $\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] \neq \boldsymbol{\Sigma}$, the raw maximum likelihood estimator is systematically downward biased in finite samples. This occurs because calculating variations against the *sample mean* rather than the *true population mean* artificially minimizes the measured distances, underrepresenting the real spatial scattering of the system.

### Bessel's Correction Elimination of Bias

To eliminate this geometric bias and yield an unbiased estimator ($\mathbf{S}$), we scale the maximum likelihood matrix by the reciprocal of the bias factor, $\frac{n}{n-1}$:

$$\mathbf{S} = \frac{n}{n-1} \widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{n}{n-1} \left( \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T \right) = \frac{1}{n-1} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$$

Evaluating the expectation confirms its lack of bias:

$$\mathbb{E}[\mathbf{S}] = \frac{n}{n-1} \mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{n}{n-1} \left(\frac{n-1}{n}\right)\boldsymbol{\Sigma} = \boldsymbol{\Sigma}$$

## 2. Operational Error Diagnostics

In structural health monitoring, our hypotheses are framed around baseline stability:
* $H_0$ (Null Hypothesis): The structural asset is completely nominal and healthy.
* $H_1$ (Alternative Hypothesis): The asset has suffered degradation, a localized crack, or structural shift.

### Statistical Error Definitions
* **Type I Error ($\alpha$):** A **False Alarm**. This occurs if the diagnostic system concludes that the structure is damaged (rejects $H_0$) when, in reality, the physical asset is perfectly nominal and healthy.
* **Type II Error ($\beta$):** A **Missed Detection**. This occurs if a structural shift or failure has compromised the asset (the true state is $H_1$), but the diagnostic platform fails to alert the team (fails to reject $H_0$).

### Consequence of an Ultra-Conservative Threshold ($\alpha = 0.0001$)

When the significance threshold $\alpha$ is reduced dramatically to eliminate false alarms, the critical boundary of the test statistic is pushed deep into the extreme tails of the baseline distribution.

1.  **Impact on Statistical Power ($1-\beta$):** Because the threshold required to trigger an anomaly alert becomes incredibly stringent, minor or emerging structural shifts under $H_1$ will fail to cross the new partition line. As a result, $\beta$ (missed detections) will sharply escalate, causing the **statistical power ($1-\beta$) of the model to collapse**. The system becomes virtually blind to early-stage material degradation.
2.  **Impact on Confidence Ellipsoids:** The confidence level of the multi-dimensional boundary scales as $1-\alpha = 0.9999$. In Mahalanobis distance space, the **geometric volume of the "healthy operational" ellipsoid expands significantly**. The boundary swells outward to encompass almost all random environmental variation, meaning the multivariate system requires an extreme, catastrophic structural shift before a data point falls outside the baseline ellipsoid to trigger a detection.

Part B: Theoretical Extension (Asymptotic Distributions & Slutsky's Theorem)

## 1. Slutsky's Theorem

**Slutsky's Theorem** is a fundamental result in asymptotic statistics that describes the convergence in distribution of combinations of random vectors.

Let $\mathbf{X}_n$ and $\mathbf{Y}_n$ be sequences of random vectors in $\mathbb{R}^m$, and let $\mathbf{A}_n$ be a sequence of random matrices in $\mathbb{R}^{m \times m}$. If:
1. $\mathbf{X}_n$ converges in distribution to a random vector $\mathbf{X}$ as $n \to \infty$ ($\mathbf{X}_n \xrightarrow{d} \mathbf{X}$)
2. $\mathbf{Y}_n$ converges in probability to a constant vector $\mathbf{c}$ as $n \to \infty$ ($\mathbf{Y}_n \xrightarrow{p} \mathbf{c}$)
3. $\mathbf{A}_n$ converges in probability to a constant matrix $\mathbf{A}$ as $n \to \infty$ ($\mathbf{A}_n \xrightarrow{p} \mathbf{A}$)

Then the following asymptotic relationships hold:
* **Additive Stability:** $\mathbf{X}_n + \mathbf{Y}_n \xrightarrow{d} \mathbf{X} + \mathbf{c}$
* **Multiplicative Stability (Matrix-Vector):** $\mathbf{A}_n \mathbf{X}_n \xrightarrow{d} \mathbf{A} \mathbf{X}$

## 2. Proof of Substitution and Operationalization

By the **Lindeberg-Lévy Multivariate Central Limit Theorem (CLT)**, the standardized empirical mean converges to a multivariate normal distribution:

$$\mathbf{Z}_n = \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \xrightarrow{d} \mathbf{Z} \sim \mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma})$$

By the **Weak Law of Large Numbers (WLLN)**, the sample covariance matrix $\mathbf{S}$ (as well as $\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}$) is a consistent estimator of the true population covariance matrix $\boldsymbol{\Sigma}$:

$$\mathbf{S} \xrightarrow{p} \boldsymbol{\Sigma} \quad \text{as } n \to \infty$$

Because matrix inversion and the matrix square root (via spectral decomposition or Cholesky factorization) are continuous mappings for positive-definite matrices, the Continuous Mapping Theorem (CMT) implies:

$$\mathbf{S}^{-1/2} \xrightarrow{p} \boldsymbol{\Sigma}^{-1/2}$$

We now form our test statistic by scaling the CLT convergence vector $\mathbf{Z}_n$ by the inverse sample standard deviation matrix $\mathbf{S}^{-1/2}$:

$$\mathbf{W}_n = \mathbf{S}^{-1/2} \mathbf{Z}_n = \mathbf{S}^{-1/2} \left[ \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right]$$

Applying **Slutsky's Theorem** directly by mapping $\mathbf{A}_n = \mathbf{S}^{-1/2}$ and $\mathbf{X}_n = \mathbf{Z}_n$:

$$\mathbf{W}_n = \mathbf{S}^{-1/2} \mathbf{Z}_n \xrightarrow{d} \boldsymbol{\Sigma}^{-1/2} \mathbf{Z}$$

Since $\mathbf{Z} \sim \mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma})$, linear transformations of Gaussian vectors yield:

$$\boldsymbol{\Sigma}^{-1/2} \mathbf{Z} \sim \mathscr{N}\left(\mathbf{0}, \boldsymbol{\Sigma}^{-1/2} \boldsymbol{\Sigma} (\boldsymbol{\Sigma}^{-1/2})^T\right) = \mathscr{N}(\mathbf{0}, \mathbf{I}_m)$$

Therefore, the studentized statistic converges to a standard multivariate normal distribution, completely free of latent parameters:

$$\mathbf{S}^{-1/2} \left[ \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right] \xrightarrow{d} \mathscr{N}(\mathbf{0}, \mathbf{I}_m)$$

### Asymptotic Dashboard Operationalization

For large engineering sample sizes $n$, we can un-standardize this result. Multiplying the standard normal distribution back by the scaled sample covariance matrix $\frac{1}{\sqrt{n}}\mathbf{S}^{1/2}$ and shifting by the local sample mean yields the operational parametric distribution utilized by the monitoring dashboard:

$$\widehat{\boldsymbol{\mu}}_n \sim \mathscr{N}\left(\boldsymbol{\mu}, \frac{1}{n}\mathbf{S}\right)$$

In practice, when running real-time anomaly detection under the null state hypothesis, the historical steady-state parameters replace the latent parameters, allowing the system to construct its confidence boundaries using:

$$\widehat{\boldsymbol{\mu}}_n \sim \mathscr{N}\left(\widehat{\boldsymbol{\mu}}_n, \frac{1}{n}\mathbf{S}\right)$$

Part C: Numerical Verification


```python
import numpy as np
import pandas as pd
from scipy import stats
from statsmodels.multivariate.manova import MANOVA

# -------------------------------------------------------------------------
# 1. SEED & SYNTHETIC DATA GENERATION (As specified by user scenario)
# -------------------------------------------------------------------------
np.random.seed(42)
n = 5000
m = 3
g = 5
chunk_size = 1000

# Generate a weak i.i.d. baseline matrix (No mean shifts injected)
data = np.random.multivariate_normal(
    mean=np.zeros(m),
    cov=np.array([[1.0, 0.4, 0.2],
                  [0.4, 1.0, 0.3],
                  [0.2, 0.3, 1.0]]),
    size=n
)

# Construct DataFrame and partition into g=5 consecutive, non-overlapping blocks
df = pd.DataFrame(data, columns=[f'sensor_{i+1}' for i in range(m)])
df['block_id'] = np.repeat(np.arange(1, g + 1), chunk_size)

# -------------------------------------------------------------------------
# 2. FROM-SCRATCH MANOVA MATRICES & BARTLETT TRANSFORMATION
# -------------------------------------------------------------------------
def run_manova_diagnostics(dataframe, dependent_cols, group_col):
    """
    Computes Wilks' Lambda, applies Bartlett's chi-squared approximation,
    and assesses first-moment homogeneity.
    """
    total_mean = dataframe[dependent_cols].mean().values

    # Initialize Within-group (W) and Between-group (B) Sum of Squares and Cross-Products matrices
    W = np.zeros((m, m))
    B = np.zeros((m, m))

    grouped = dataframe.groupby(group_col)

    for _, group in grouped:
        group_data = group[dependent_cols].values
        group_mean = group_data.mean(axis=0)
        n_j = len(group_data)

        # Within-group SSCP accumulation
        dev_W = group_data - group_mean
        W += dev_W.T @ dev_W

        # Between-group SSCP accumulation
        dev_B = group_mean - total_mean
        B += n_j * np.outer(dev_B, dev_B)

    # Calculate Wilks' Lambda using determinants
    det_W = np.linalg.det(W)
    det_T = np.linalg.det(B + W)
    wilks_lambda = det_W / det_T

    # Apply Bartlett's Chi-Squared Approximation
    # Scaling factor multiplier
    multiplier = -((n - 1) - (m + g) / 2.0)
    chi2_calc = multiplier * np.log(wilks_lambda)

    # Degrees of freedom for the tracking distribution
    dof = m * (g - 1)
    p_value = 1.0 - stats.chi2.cdf(chi2_calc, dof)

    return wilks_lambda, chi2_calc, dof, p_value

# Execute the custom diagnostic tracker
w_lambda, chi2_stat, df_dof, p_val = run_manova_diagnostics(df, [f'sensor_{i+1}' for i in range(m)], 'block_id')

# -------------------------------------------------------------------------
# 3. VERIFICATION VIA STATSMODELS
# -------------------------------------------------------------------------
formula = ' + '.join(df.columns[:-1]) + ' ~ C(block_id)'
manova_model = MANOVA.from_formula(formula, data=df)
manova_results = manova_model.mv_test()

# Extract parameters from statsmodels dictionary structure for confirmation
sm_lambda = manova_results.results['C(block_id)']['stat'].iloc[0, 0]

# -------------------------------------------------------------------------
# 4. REPORTING LOGS
# -------------------------------------------------------------------------
print("=" * 65)
print("             MANOVA FIRST-MOMENT HOMOGENEITY REPORT       ")
print("=" * 65)
print(f"Calculated Wilks' Lambda (Λ):      {w_lambda:.6f}")
print(f"Statsmodels Verified Lambda:      {sm_lambda:.6f}")
print(f"Bartlett Chi-Square (χ²_calc):    {chi2_stat:.4f}")
print(f"Degrees of Freedom (m * (g - 1)): {df_dof}")
print(f"Asymptotic p-value:               {p_val:.6f}")
print("-" * 65)

alpha = 0.05
if p_val < alpha:
    print(f"CONCLUSION: Reject H0 at α = {alpha}.")
    print("Diagnostic Warning: The baseline strain center has significantly shifted over time.")
    print("The First Moment Homogeneity condition is VIOLATED. Do not initialize global PCA.")
else:
    print(f"CONCLUSION: Fail to Reject H0 at α = {alpha}.")
    print("Diagnostic Pass: No statistically significant mean shift detected across the blocks.")
    print("The joint mean remains CONSTANT. The weak i.i.d. assumption holds for PCA/FA.")
print("=" * 65)
```

Part A: Theoretical Fundamentals (Coordinate Projections & Orthogonality)

## 1. Orthogonal Projections and Decorrelation

### Step-by-Step Proof of Covariance Diagonalization
We are given the centered vector $\widetilde{\mathbf{X}}_i = \mathbf{X}_i - \boldsymbol{\mu}$ and its population covariance matrix $\boldsymbol{\Sigma} = \mathbb{E}[\widetilde{\mathbf{X}}_i \widetilde{\mathbf{X}}_i^T]$. The spectral decomposition of this symmetric, positive-semidefinite matrix is:

$$\boldsymbol{\Sigma} = \mathbf{P} \mathbf{\Lambda} \mathbf{P}^T$$

where $\mathbf{P}$ is an $m \times m$ orthogonal matrix ($\mathbf{P}^T \mathbf{P} = \mathbf{P}\mathbf{P}^T = \mathbf{I}_m$) containing the eigenvectors as columns, and $\mathbf{\Lambda}$ is a diagonal matrix containing the eigenvalues $\lambda_1, \lambda_2, \dots, \lambda_m$.

We define the transformed coordinate vector as $\mathbf{Z}_i = \mathbf{P}^T \widetilde{\mathbf{X}}_i$. To find its theoretical covariance matrix, we evaluate the expectation of its outer product:

$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbb{E}\left[ (\mathbf{P}^T \widetilde{\mathbf{X}}_i) (\mathbf{P}^T \widetilde{\mathbf{X}}_i)^T \right]$$

Applying the matrix transpose property $(\mathbf{A}\mathbf{B})^T = \mathbf{B}^T \mathbf{A}^T$:

$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbb{E}\left[ \mathbf{P}^T \widetilde{\mathbf{X}}_i \widetilde{\mathbf{X}}_i^T \mathbf{P} \right]$$

Because the expectation operator $\mathbb{E}[\cdot]$ is linear and $\mathbf{P}$ is a deterministic matrix derived from the population parameters, we pull $\mathbf{P}^T$ and $\mathbf{P}$ outside the expectation:

$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{P}^T \mathbb{E}[\widetilde{\mathbf{X}}_i \widetilde{\mathbf{X}}_i^T] \mathbf{P}$$

Substituting the definition of the baseline covariance matrix $\boldsymbol{\Sigma}$:

$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{P}^T \boldsymbol{\Sigma} \mathbf{P}$$

Next, we substitute the spectral decomposition $\boldsymbol{\Sigma} = \mathbf{P} \mathbf{\Lambda} \mathbf{P}^T$ into the expression:

$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{P}^T \left( \mathbf{P} \mathbf{\Lambda} \mathbf{P}^T \right) \mathbf{P}$$

Using the associativity of matrix multiplication:

$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \left(\mathbf{P}^T \mathbf{P}\right) \mathbf{\Lambda} \left(\mathbf{P}^T \mathbf{P}\right)$$

Since $\mathbf{P}$ is orthogonal, $\mathbf{P}^T \mathbf{P} = \mathbf{I}_m$. Substituting the identity matrix yields:

$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{I}_m \mathbf{\Lambda} \mathbf{I}_m = \mathbf{\Lambda}$$

$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \text{diag}(\lambda_1, \lambda_2, \dots, \lambda_m)$$

### Geometric and Statistical Interpretation
* **Statistical Meaning:** Because $\mathbf{\Lambda}$ is strictly diagonal, the cross-covariance between any two distinct transformed coordinates $Z_{i,j}$ and $Z_{i,k}$ (where $j \neq k$) is exactly zero:
  
  $$\text{Cov}(Z_{i,j}, Z_{i,k}) = \mathbb{E}[Z_{i,j} Z_{i,k}] = 0$$
  
  This demonstrates that the linear transformations completely **decorrelate** the sensor channels. Information redundancy across channels is removed.
* **Geometric Meaning:** The orthogonal matrix $\mathbf{P}^T$ represents a rigid **rotation** of the coordinate axes. It shifts the original, entangled coordinate frame of the sensor network into a new alignment along the principal axes of the data dispersion ellipsoid.

## 2. Variance Preservation and Subspace Partitioning

### Mathematical Proof of Total Variance Conservation
The total variance of a system is the sum of the variances of its individual components, which corresponds to the trace of its covariance matrix. To show that this quantity is conserved under rotation, we evaluate $\text{tr}(\mathbf{\Lambda})$:

$$\text{tr}(\mathbf{\Lambda}) = \text{tr}(\mathbf{P}^T \boldsymbol{\Sigma} \mathbf{P})$$

The trace operator possesses a **cyclic permutation invariance property**, stating that for any matrices $\mathbf{A}$ and $\mathbf{B}$ of matching dimensions such that the products are square, $\text{tr}(\mathbf{A}\mathbf{B}) = \text{tr}(\mathbf{B}\mathbf{A})$. Grouping our terms as $\mathbf{A} = \mathbf{P}^T \boldsymbol{\Sigma}$ and $\mathbf{B} = \mathbf{P}$, we cyclically permute the matrix product:

$$\text{tr}((\mathbf{P}^T \boldsymbol{\Sigma}) \mathbf{P}) = \text{tr}(\mathbf{P} (\mathbf{P}^T \boldsymbol{\Sigma}))$$

Using associativity to group the inner products:

$$\text{tr}(\mathbf{\Lambda}) = \text{tr}((\mathbf{P} \mathbf{P}^T) \boldsymbol{\Sigma})$$

Since $\mathbf{P}$ is orthogonal, its external product is also an identity matrix ($\mathbf{P}\mathbf{P}^T = \mathbf{I}_m$):

$$\text{tr}(\mathbf{\Lambda}) = \text{tr}(\mathbf{I}_m \boldsymbol{\Sigma}) = \text{tr}(\boldsymbol{\Sigma})$$

Because $\mathbf{\Lambda}$ is diagonal, its trace is simply the scalar sum of its diagonal entries:

$$\text{tr}(\boldsymbol{\Sigma}) = \text{tr}(\mathbf{\Lambda}) = \sum_{j=1}^m \lambda_j$$

This proves that rotating the coordinate system reallocates variance among the axes but leaves the total energy of the multivariate system invariant.

### Subspace Ratio Formulations
When partitioning the empirical domain into an active, explained principal subspace of $k$ components and an omitted, unexplained residual subspace of $m-k$ components, the performance ratios are defined explicitly using the sorted sample eigenvalues $\hat{\lambda}_1 \ge \hat{\lambda}_2 \ge \dots \ge \hat{\lambda}_m$:

* **Cumulative Explained Variance Ratio $\Phi(k)$:** Measures the proportion of total information preserved within the active tracking layer.
  
  $$\Phi(k) = \frac{\text{tr}(\widehat{\mathbf{P}}_k^T \mathbf{S} \widehat{\mathbf{P}}_k)}{\text{tr}(\mathbf{S})} = \frac{\sum_{j=1}^k \hat{\lambda}_j}{\sum_{j=1}^m \hat{\lambda}_j}$$

* **Residual Unexplained Variance Ratio $\Psi(k)$:** Measures the discarded energy proportion left in the noise floor workspace.
  
  $$\Psi(k) = \frac{\text{tr}(\widehat{\mathbf{P}}_{m-k}^T \mathbf{S} \widehat{\mathbf{P}}_{m-k})}{\text{tr}(\mathbf{S})} = \frac{\sum_{j=k+1}^m \hat{\lambda}_j}{\sum_{j=1}^m \hat{\lambda}_j} = 1 - \Phi(k)$$

## 3. Subspace Decomposition and Diagnostic Mapping

### Vector Definitions
Using the empirical projection matrix partitioned into $\widehat{\mathbf{P}}_k \in \mathbb{R}^{m \times k}$ and $\widehat{\mathbf{P}}_{m-k} \in \mathbb{R}^{m \times (m-k)}$, the original vector components are $\mathbf{z}_{i,k} = \widehat{\mathbf{P}}_k^T (\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n)$ and $\mathbf{z}_{i,m-k} = \widehat{\mathbf{P}}_{m-k}^T (\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n)$.
* **Reconstructed Vector ($\widehat{\mathbf{x}}_i$):** The projection of the observation back into the $k$-dimensional principal subspace:
  
  $$\widehat{\mathbf{x}}_i = \widehat{\boldsymbol{\mu}}_n + \widehat{\mathbf{P}}_k \mathbf{z}_{i,k}$$

* **Residual Error Vector ($\mathbf{e}_i$):** The portion of the signal projected onto the orthogonal complement space:
  
  $$\mathbf{e}_i = (\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n) - \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} = \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}$$

### Pythagorean Subspace Energy Proof
To show the clean breakdown of the squared Euclidean norm, we expand the vector definition:

$$\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = (\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n)^T (\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n)$$

Substituting the partitioned subspace components $(\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n) = \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}$:

$$\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = \left( \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k} \right)^T \left( \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k} \right)$$

$$\begin{aligned}
= \;& \mathbf{z}_{i,k}^T \widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \mathbf{z}_{i,k}^T \widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k} \\
& + \mathbf{z}_{i,m-k}^T \widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \mathbf{z}_{i,m-k}^T \widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}
\end{aligned}$$

Because the full eigenvector matrix $\widehat{\mathbf{P}}$ is orthogonal, its sub-blocks are mutually orthogonal ($\widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_{m-k} = \mathbf{0}_{k \times (m-k)}$ and $\widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_k = \mathbf{0}_{(m-k) \times k}$). This forces the two cross-product terms to vanish:

$$\mathbf{z}_{i,k}^T \left( \widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_{m-k} \right) \mathbf{z}_{i,m-k} = 0, \quad \mathbf{z}_{i,m-k}^T \left( \widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_k \right) \mathbf{z}_{i,k} = 0$$

Furthermore, the individual sub-blocks preserve length on their respective subspaces ($\widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_k = \mathbf{I}_k$ and $\widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_{m-k} = \mathbf{I}_{m-k}$). Substituting these identities into the remaining terms yields:

$$\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = \mathbf{z}_{i,k}^T \mathbf{I}_k \mathbf{z}_{i,k} + \mathbf{z}_{i,m-k}^T \mathbf{I}_{m-k} \mathbf{z}_{i,m-k} = \mathbf{z}_{i,k}^T \mathbf{z}_{i,k} + \mathbf{z}_{i,m-k}^T \mathbf{z}_{i,m-k}$$

$$\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = \|\mathbf{z}_{i,k}\|^2 + \|\mathbf{z}_{i,m-k}\|^2$$

### SHM Diagnostic Anomaly Matching

Hotelling's $T^2$ and the residual $Q$ statistic target distinct geometric regions of the system's multivariate space:

* **Hotelling’s $T^2$ Statistic (In-Subspace Distance):** Measures the Mahalanobis distance within the principal component plane:
    
    $$T^2_i = \mathbf{z}_{i,k}^T \mathbf{\Lambda}_k^{-1} \mathbf{z}_{i,k} = \sum_{j=1}^k \frac{z_{i,j}^2}{\hat{\lambda}_j}$$
    
    It tracks changes in the magnitude of variations along the expected, healthy directions established during baseline training.
* **Residual $Q$ Statistic / SPE (Out-of-Subspace Distance):** Measures the squared Euclidean length of the residual error vector:
    
    $$Q_i = \|\mathbf{e}_i\|^2 = \sum_{j=k+1}^m z_{i,j}^2$$
    
    It checks how well new data points fit the established model subspace, functioning as an orthogonal breakdown detector.

| Operational Event | Primary Matching Statistic | Structural Mechanical Justification |
| :--- | :--- | :--- |
| **Environmental Load Anomaly** *(e.g., Extreme Operational Wind)* | **Hotelling's $T^2$ Statistic** | High winds compress and flex the asset along its standard, macro-level physical degrees of freedom. The underlying structural load pathways remain intact; the system simply undergoes a large-amplitude displacement. This causes the data point to move far out along the dominant, expected baseline eigenvectors, driving a spike in $T^2$ while $Q$ stays nominal. |
| **Internal Structural Fracture** *(e.g., Fatigue Crack Propagation)* | **Residual $Q$ Statistic (SPE)** | A physical crack fundamentally breaks the geometry of the system and shifts its local stiffness matrix. The sensors begin capturing novel coupling relationships that were physically impossible during baseline conditions. The data vector is forced out of the established $k$-dimensional model hyperplane, causing the projection error to spike and triggering a major response in the $Q$ metric. |

Part B: Numerical Verification

```python
import numpy as np
import pandas as pd

class SHMDiagnosticsPipeline:
    def __init__(self, n_sensors: int = 4):
        self.m = n_sensors
        self.mu_hat = None
        self.S = None
        self.eigenvalues = None
        self.eigenvectors = None

    def fit_baseline_and_evaluate_truncation(self, X: np.ndarray) -> dict:
        """
        Fits the baseline model and evaluates the empirical expectations of
        Hotelling's T^2 and residual Q statistics across all possible truncation thresholds.

        Parameters:
        -----------
        X : np.ndarray of shape (n, m)
            The raw snapshot baseline matrix containing n observations and m sensor channels.

        Returns:
        --------
        results : dict
            A nested dictionary mapping each truncation boundary k to its corresponding
            empirical expected value of T^2 and Q.
        """
        n, m = X.shape
        assert m == self.m, f"Expected {self.m} sensors, but matrix contains {m} columns."

        # -------------------------------------------------------------------------
        # 1. STANDARDIZE AND CENTER THE INCOMING REALIZATIONS
        # -------------------------------------------------------------------------
        # Calculate empirical sample mean vector (1 x m)
        self.mu_hat = np.mean(X, axis=0)
        # Center the dataset
        X_centered = X - self.mu_hat

        # -------------------------------------------------------------------------
        # 2. DECOMPOSE UNBIASED COVARIANCE MATRIX & SORT IN DESCENDING ORDER
        # -------------------------------------------------------------------------
        # Calculate the unbiased sample covariance matrix S (using Bessel's correction: n-1)
        self.S = (X_centered.T @ X_centered) / (n - 1)

        # Perform spectral decomposition. eigh is optimized for symmetric/Hermitian matrices
        evals, evecs = np.linalg.eigh(self.S)

        # np.linalg.eigh returns components in ascending order. Reverse arrays for descending order.
        idx = np.argsort(evals)[::-1]
        self.eigenvalues = evals[idx]
        self.eigenvectors = evecs[:, idx]

        # Project all centered snapshots onto the full orthogonal principal component space
        # Z matrix shape: (n x m)
        Z = X_centered @ self.eigenvectors

        # -------------------------------------------------------------------------
        # 3. LOOP THROUGH EVERY POSSIBLE SUBSPACE CUTOFF VALUE k in {1, 2, 3}
        # -------------------------------------------------------------------------
        results = {}
        for k in range(1, m):
            # Extract relevant principal components and eigenvalues for the active subspace
            Z_k = Z[:, :k]
            lambda_k = self.eigenvalues[:k]

            # Extract residual components for the orthogonal complement workspace
            Z_residual = Z[:, k:]

            # 4. COMPUTE STATISTIC ARRAYS AND THEIR EMPIRICAL EXPECTED VALUES
            # Hotelling's T^2 calculation for each snapshot: sum_{j=1}^k (z_{i,j}^2 / lambda_j)
            t2_samples = np.sum((Z_k ** 2) / lambda_k, axis=1)
            E_T2 = np.mean(t2_samples)

            # Residual Q statistic calculation for each snapshot: sum_{j=k+1}^m (z_{i,j}^2)
            q_samples = np.sum(Z_residual ** 2, axis=1)
            E_Q = np.mean(q_samples)

            # Store calculations mapped to the specific truncation boundary
            results[k] = {
                "Expected_T2": E_T2,
                "Expected_Q": E_Q
            }

        return results

# -------------------------------------------------------------------------
# EXECUTION AND VERIFICATION VERIFICATION WITH SYNTHETIC DATA
# -------------------------------------------------------------------------
if __name__ == "__main__":
    np.random.seed(42)
    n_samples = 3000
    n_sensors = 4

    # Generate nominal synthetic data with known baseline cross-covariance structure
    true_cov = np.array([[1.2, 0.5, 0.3, 0.1],
                         [0.5, 1.0, 0.4, 0.2],
                         [0.3, 0.4, 0.8, 0.3],
                         [0.1, 0.2, 0.3, 0.6]])

    X_synthetic = np.random.multivariate_normal(mean=np.zeros(n_sensors), cov=true_cov, size=n_samples)

    # Initialize optimization pipeline
    pipeline = SHMDiagnosticsPipeline(n_sensors=n_sensors)
    metrics = pipeline.fit_baseline_and_evaluate_truncation(X_synthetic)

    # Print out summary evaluations to verify theoretical constraints
    print("=" * 65)
    print("        SHM SUBSPACE OPTIMIZATION SUMMARY (n=3000, m=4)      ")
    print("=" * 65)
    print(f"{'Cutoff (k)':<12} | {'Empirical E[T²]':<20} | {'Empirical E[Q]':<20}")
    print("-" * 65)
    for k, val in metrics.items():
        print(f"{k:<12} | {val['Expected_T2']:<20.4f} | {val['Expected_Q']:<20.4f}")
    print("=" * 65)
    print("Theoretical Note:")
    print(" - For any cutoff k, E[T²] converges asymptotically to exactly k.")
    print(" - For any cutoff k, E[Q] matches the sum of the remaining eigenvalues.")
    print(f"   Sum of all sample eigenvalues: {np.sum(pipeline.eigenvalues):.4f}")
    print("=" * 65)
```

Latent Subspace Decomposition via Factor Analysis (FA)

Part A: Theoretical Fundamentals (Generative Model & Commonalities)

## 1. The Fundamental Equation of Factor Analysis

### Step-by-Step Mathematical Proof
We begin with the linear generative structural model for the normalized, zero-mean baseline feature vector $\mathbf{Z}_i \in \mathbb{R}^m$:

$$\mathbf{Z}_i = \boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i$$

By definition, the universal population correlation matrix $\mathbf{R}$ is the expected value of the outer product of the centered vector $\mathbf{Z}_i$:

$$\mathbf{R} = \mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T]$$

Substituting the generative structural model expression into the outer product yields:

$$\mathbf{R} = \mathbb{E}\left[ (\boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i) (\boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i)^T \right]$$

Applying the transpose operator property $(\mathbf{A} + \mathbf{B})^T = \mathbf{A}^T + \mathbf{B}^T$ and expanding the terms gives:

$$\mathbf{R} = \mathbb{E}\left[ (\boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i) (\mathbf{F}_i^T \boldsymbol{\Lambda}^T + \boldsymbol{\epsilon}_i^T) \right]$$

$$= \mathbb{E}\left[ \boldsymbol{\Lambda} \mathbf{F}_i \mathbf{F}_i^T \boldsymbol{\Lambda}^T + \boldsymbol{\Lambda} \mathbf{F}_i \boldsymbol{\epsilon}_i^T + \boldsymbol{\epsilon}_i \mathbf{F}_i^T \boldsymbol{\Lambda}^T + \boldsymbol{\epsilon}_i \boldsymbol{\epsilon}_i^T \right]$$

By utilizing the linearity of the expectation operator $\mathbb{E}[\cdot]$, we can evaluate each structural matrix term individually:

$$\mathbf{R} = \boldsymbol{\Lambda} \mathbb{E}[\mathbf{F}_i \mathbf{F}_i^T] \boldsymbol{\Lambda}^T + \boldsymbol{\Lambda} \mathbb{E}[\mathbf{F}_i \boldsymbol{\epsilon}_i^T] + \mathbb{E}[\boldsymbol{\epsilon}_i \mathbf{F}_i^T] \boldsymbol{\Lambda}^T + \mathbb{E}[\boldsymbol{\epsilon}_i \boldsymbol{\epsilon}_i^T]$$

We now substitute the standard latent constraint parameters into the expanded equation:
1. **Orthogonal Common Factors:** $\mathbb{E}[\mathbf{F}_i \mathbf{F}_i^T] = \mathbf{I}_{k \times k}$
2. **Mutual Latent Independence:** $\mathbb{E}[\boldsymbol{\epsilon}_i \mathbf{F}_i^T] = \mathbf{0}_{m \times k}$ (which also implies its transpose $\mathbb{E}[\mathbf{F}_i \boldsymbol{\epsilon}_i^T] = \mathbf{0}_{k \times m}$)
3. **Diagonalized Specific Variance:** $\mathbb{E}[\boldsymbol{\epsilon}_i \boldsymbol{\epsilon}_i^T] = \boldsymbol{\Psi}$

Substituting these constraints yields the **Fundamental Equation of Factor Analysis**:

$$\mathbf{R} = \boldsymbol{\Lambda} \mathbf{I}_{k \times k} \boldsymbol{\Lambda}^T + \boldsymbol{\Lambda} (\mathbf{0}) + (\mathbf{0}) \boldsymbol{\Lambda}^T + \boldsymbol{\Psi}$$

$$\mathbf{R} = \boldsymbol{\Lambda}\boldsymbol{\Lambda}^T + \boldsymbol{\Psi}$$

### Parameter Decompositions and Diagnostic Interpretations
Focusing directly on the $j$-th diagonal element of the correlation matrix $\mathbf{R}$ (where $R_{jj} = 1$ due to normalization), the expression simplifies to:

$$1 = \sum_{l=1}^k \lambda_{jl}^2 + \varphi_j^2$$

* **Communality ($h_j^2 = \sum_{l=1}^k \lambda_{jl}^2$):** The proportion of the $j$-th sensor's variance that is explained by the shared, global structural dynamics of the asset.
    * *Diagnostic Meaning:* A high communality means the sensor is tightly integrated into the baseline structural network, making it highly effective for detecting macro-level operational shifts.
* **Uniqueness / Specific Variance ($\varphi_j^2$):** The proportion of the $j$-th sensor's variance that cannot be explained by the shared components. This represents localized white noise or channel-specific variations.
    * *Diagnostic Meaning:* If a sensor's uniqueness suddenly increases while others remain nominal, it indicates a localized issue—such as an isolated instrument fault or a loose mounting bracket—rather than a global structural failure.

---

## 2. Factor Rotations and Indeterminacy

### Conceptual Divergence: PCA vs. Varimax-Rotated Loading Vectors
* **Raw PCA Loading Vectors:** Principal component loadings maximize the total variance explained sequentially. The first column accounts for the largest possible variation, the second maximizes the remaining variation, and so on. This often results in a complex, un-aligned matrix where many sensors display moderate loadings across multiple components, clouding physical interpretability.
* **Varimax-Rotated Factor Loading Vectors:** Varimax rotation modifies the coordinate system to maximize the variance of the *squared loadings* within each column:
    
    $$V = \frac{1}{m} \sum_{l=1}^k \left[ m \sum_{j=1}^m (\lambda_{jl}^2)^2 - \left( \sum_{j=1}^m \lambda_{jl}^2 \right)^2 \right]$$
    
    This rotation drives the values toward a **simple structure**, forcing loadings to be either very high or near zero. Conceptually, it maps the abstract mathematical axes of PCA to specific physical structural sub-assemblies (e.g., separating bending modes from torsional modes).

### Invariance Proof of Communality and Global Approximation
Let $\mathbf{T}$ be an orthogonal rotation matrix satisfying $\mathbf{T}\mathbf{T}^T = \mathbf{T}^T\mathbf{T} = \mathbf{I}_{k \times k}$. We transform the original loading matrix into a rotated configuration $\boldsymbol{\Lambda}^* = \boldsymbol{\Lambda}\mathbf{T}$.

To see why the global covariance approximation remains completely unchanged, we substitute the rotated loadings back into the fundamental equation:

$$\mathbf{R}^* = \boldsymbol{\Lambda}^*(\boldsymbol{\Lambda}^*)^T + \boldsymbol{\Psi}$$

$$\mathbf{R}^* = (\boldsymbol{\Lambda}\mathbf{T})(\boldsymbol{\Lambda}\mathbf{T})^T + \boldsymbol{\Psi}$$

Applying the transpose matrix product rule $(\mathbf{A}\mathbf{B})^T = \mathbf{B}^T \mathbf{A}^T$:

$$\mathbf{R}^* = \boldsymbol{\Lambda}(\mathbf{T}\mathbf{T}^T)\boldsymbol{\Lambda}^T + \boldsymbol{\Psi}$$

Since $\mathbf{T}$ is orthogonal, the internal product resolves to the identity matrix ($\mathbf{T}\mathbf{T}^T = \mathbf{I}_k$):

$$\mathbf{R}^* = \boldsymbol{\Lambda}\mathbf{I}_k\boldsymbol{\Lambda}^T + \boldsymbol{\Psi} = \boldsymbol{\Lambda}\boldsymbol{\Lambda}^T + \boldsymbol{\Psi} = \mathbf{R}$$

Because the global product $\boldsymbol{\Lambda}\boldsymbol{\Lambda}^T$ is invariant to rotation, its diagonal entries remain exactly the same. The communality of the $j$-th sensor ($h_j^2 = [\boldsymbol{\Lambda}\boldsymbol{\Lambda}^T]_{jj}$) is preserved, proving that orthogonal rotations merely reallocate coordinate assignments among the latent factors without altering the physical integrity of the underlying model.

## 3. Latent Factor Scoring via Thomson's Regression Method

### Derivation of the Joint Partitioned Covariance Matrix $\boldsymbol{\Sigma}_{YY}$
We construct the joint partitioned vector containing both observable features and unobservable components:

$$\mathbf{Y}_i = \begin{bmatrix} \mathbf{Z}_i \\ \mathbf{F}_i \end{bmatrix} \in \mathbb{R}^{m + k}$$

Given that the components are zero-mean, the joint covariance matrix $\boldsymbol{\Sigma}_{YY}$ is defined by the block matrix expansion:

$$\boldsymbol{\Sigma}_{YY} = \mathbb{E}[\mathbf{Y}_i \mathbf{Y}_i^T] = \mathbb{E}\left[ \begin{bmatrix} \mathbf{Z}_i \\ \mathbf{F}_i \end{bmatrix} \begin{bmatrix} \mathbf{Z}_i^T & \mathbf{F}_i^T \end{bmatrix} \right] = \begin{bmatrix} \mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] & \mathbb{E}[\mathbf{Z}_i \mathbf{F}_i^T] \\ \mathbb{E}[\mathbf{F}_i \mathbf{Z}_i^T] & \mathbb{E}[\mathbf{F}_i \mathbf{F}_i^T] \end{bmatrix} = \begin{bmatrix} \boldsymbol{\Sigma}_{11} & \boldsymbol{\Sigma}_{12} \\ \boldsymbol{\Sigma}_{21} & \boldsymbol{\Sigma}_{22} \end{bmatrix}$$

We evaluate each sub-block individually to find the expanded parameters:
* $\boldsymbol{\Sigma}_{11} = \mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{R}$ (The primary correlation matrix)
* $\boldsymbol{\Sigma}_{22} = \mathbb{E}[\mathbf{F}_i \mathbf{F}_i^T] = \mathbf{I}_{k \times k}$ (The orthogonal factor matrix constraint)

For the cross-covariance block $\boldsymbol{\Sigma}_{12}$, we substitute the generative structural model $\mathbf{Z}_i = \boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i$:

$$\boldsymbol{\Sigma}_{12} = \mathbb{E}[\mathbf{Z}_i \mathbf{F}_i^T] = \mathbb{E}[(\boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i) \mathbf{F}_i^T] = \boldsymbol{\Lambda}\mathbb{E}[\mathbf{F}_i \mathbf{F}_i^T] + \mathbb{E}[\boldsymbol{\epsilon}_i \mathbf{F}_i^T]$$

Using the constraints $\mathbb{E}[\mathbf{F}_i \mathbf{F}_i^T] = \mathbf{I}_k$ and $\mathbb{E}[\boldsymbol{\epsilon}_i \mathbf{F}_i^T] = \mathbf{0}$:

$$\boldsymbol{\Sigma}_{12} = \boldsymbol{\Lambda}\mathbf{I}_k + \mathbf{0} = \boldsymbol{\Lambda}$$

Finally, the remaining block is simply the transpose of the cross-covariance block:

$$\boldsymbol{\Sigma}_{21} = \mathbb{E}[\mathbf{F}_i \mathbf{Z}_i^T] = (\mathbb{E}[\mathbf{Z}_i \mathbf{F}_i^T])^T = \boldsymbol{\Lambda}^T$$

Assembling these blocks gives the complete joint population covariance matrix:

$$\boldsymbol{\Sigma}_{YY} = \begin{bmatrix} \mathbf{R} & \boldsymbol{\Lambda} \\ \boldsymbol{\Lambda}^T & \mathbf{I}_{k \times k} \end{bmatrix}$$

### Derivation of Thomson's Estimator Equation
Under the assumption of a multivariate normal distribution, the conditional expectation of a partitioned system is given by the standard projection identity:

$$\mathbb{E}[\mathbf{X}_2 | \mathbf{x}_1] = \boldsymbol{\mu}_2 + \boldsymbol{\Sigma}_{21} \boldsymbol{\Sigma}_{11}^{-1} (\mathbf{x}_1 - \boldsymbol{\mu}_1)$$

We map our variables directly to this partitioned framework:
* $\mathbf{X}_1 = \mathbf{Z}_i$ with mean $\boldsymbol{\mu}_1 = \mathbf{0}$
* $\mathbf{X}_2 = \mathbf{F}_i$ with mean $\boldsymbol{\mu}_2 = \mathbf{0}$
* Sub-blocks: $\boldsymbol{\Sigma}_{11} = \mathbf{R}$ and $\boldsymbol{\Sigma}_{21} = \boldsymbol{\Lambda}^T$

Substituting these values into the projection identity yields the final scoring equation for Thomson's regression estimator:

$$\mathbf{f}_i = \mathbb{E}[\mathbf{F}_i | \mathbf{z}_i] = \mathbf{0} + \boldsymbol{\Lambda}^T \mathbf{R}^{-1} (\mathbf{z}_i - \mathbf{0})$$

$$\mathbf{f}_i = \boldsymbol{\Lambda}^T \mathbf{R}^{-1} \mathbf{z}_i$$

This formulation uses the global baseline correlation structures to reconstruct the real-time values of the unobservable latent factors $\mathbf{f}_i$ from the incoming sensor measurements $\mathbf{z}_i$.

Part B: Parametric Factor Partition Engine & UI Architecture

```python
import numpy as np
import pandas as pd
from sklearn.decomposition import FactorAnalysis

def run_factor_analysis_pipeline(df: pd.DataFrame, target_factors: int) -> tuple:
    """
    Executes an end-to-end Factor Analysis pipeline with Varimax rotation,
    enforcing rigorous dimensionality constraints, numerical guardrails,
    and computing Thomson latent factor scores.

    Parameters:
    -----------
    df : pd.DataFrame
        The raw input dataframe containing sensor observations.
    target_factors : int (k)
        The requested dimension of the latent structural subspace.

    Returns:
    --------
    F : np.ndarray of shape (n, k)
        The latent physical factor states computed via Thomson's regression method.
    metrics : dict
        A dictionary containing the communality, uniqueness, and global metrics.
    """
    # -------------------------------------------------------------------------
    # 1. Z-SCORE MAPPING LAYER WITH NUMERICAL GUARDRAILS
    # -------------------------------------------------------------------------
    # Isolate only the numeric observation columns
    numeric_df = df.select_dtypes(include=[np.number])
    X = numeric_df.values
    n, m = X.shape
    k = target_factors

    # Calculate empirical sample moments (ddof=1)
    means = np.mean(X, axis=0)
    stds = np.std(X, axis=0, ddof=1)

    # Numerical guardrail: Catch zero-variance channels and cap to minimum threshold
    stds_safe = np.where(stds == 0, 1e-15, stds)

    # Transform observations into standard normal space Z
    Z = (X - means) / stds_safe

    # -------------------------------------------------------------------------
    # 2. DIMENSIONALITY CONSTRAINT LAYER
    # -------------------------------------------------------------------------
    if n <= m:
        raise ValueError(
            f"Dimensionality Constraint Violated: Snapshot count n ({n}) "
            f" must be strictly greater than the channel feature count m ({m})."
        )
    if k >= m:
        raise ValueError(
            f"Subspace Cutoff Error: Requested latent space dimension k ({k}) "
            f"must be strictly less than the available feature space dimension m ({m})."
        )

    # -------------------------------------------------------------------------
    # 3. INFORMATION DECOMPOSITION LAYER (FA + VARIMAX)
    # -------------------------------------------------------------------------
    # Fit the Factor Analysis model using a randomized SVD/EM approach
    # rotation='varimax' applies the orthogonal coordinate rotation matrix T automatically
    fa = FactorAnalysis(n_components=k, rotation='varimax', random_state=42)
    fa.fit(Z)

    # Extract the Varimax-rotated factor loading matrix (Lambda)
    # Lambda shape: (m x k)
    lambda_rotated = fa.components_.T

    # Compute the Uniqueness Vector (phi^2) directly from the fitted model
    # phi^2 shape: (m,)
    uniqueness = fa.noise_variance_

    # Compute the Communality Vector (h^2)
    # h_j^2 = sum_{l=1}^k (lambda_jl^2)
    communality = np.sum(lambda_rotated ** 2, axis=1)

    # -------------------------------------------------------------------------
    # 4. MATHEMATICAL STATE RECONSTRUCTION (THOMSON'S REGRESSION METHOD)
    # -------------------------------------------------------------------------
    # Calculate the universal empirical correlation matrix R of the scaled features
    R = np.corrcoef(Z, rowvar=False)

    # Handle potential multicollinearity by adding a tiny ridge multiplier to the diagonal if needed
    if np.linalg.cond(R) > 1 / np.finfo(float).eps:
        R_inv = np.linalg.pinv(R)
    else:
        R_inv = np.linalg.inv(R)

    # Implement Thomson's MMSE equation: F = Z * R^-1 * Lambda_rotated
    # Matrix operations resolve dimensions: (n x m) x (m x m) x (m x k) -> (n x k)
    F = Z @ R_inv @ lambda_rotated

    # -------------------------------------------------------------------------
    # 5. PERFORMANCE DIAGNOSTIC TELEMETRY
    # -------------------------------------------------------------------------
    # Calculate the empirical variance of the reconstructed factors (ddof=1)
    factor_variances = np.var(F, axis=0, ddof=1)

    # Calculate system-wide percentage averages
    avg_communality_pct = np.mean(communality) * 100
    avg_uniqueness_pct = np.mean(uniqueness) * 100

    # Print a clean confirmation breakdown out to the logging console
    print("=" * 70)
    print("         FACTOR ANALYSIS ENGINE TELEMETRY REGISTRATION LOG       ")
    print("=" * 70)
    print(f"Operational Snapshots (n):       {n}")
    print(f"Isolated Sensor Channels (m):    {m}")
    print(f"Retained Latent Subspace (k):    {k}\n")
    print("-" * 70)
    print(f"Average System Communality %:    {avg_communality_pct:.3f}%")
    print(f"Average System Uniqueness %:     {avg_uniqueness_pct:.3f}%")
    print(f"Total Model Captured Energy %:   {(avg_communality_pct + avg_uniqueness_pct):.3f}%")
    print("-" * 70)
    print("Empirical Factor Score Variances (Unbiased, ddof=1):")
    for idx, var in enumerate(factor_variances):
        print(f"  └─ Factor {idx + 1} State Variance: {var:.6f}")
    print("=" * 70)

    # Compile performance dictionary metrics
    metrics = {
        "channels": numeric_df.columns.tolist(),
        "communality": communality,
        "uniqueness": uniqueness,
        "avg_communality_pct": avg_communality_pct,
        "avg_uniqueness_pct": avg_uniqueness_pct,
        "factor_variances": factor_variances,
        "loadings": lambda_rotated
    }

    return F, metrics


# -------------------------------------------------------------------------
# SCENARIO EXECUTION WRAPPER
# -------------------------------------------------------------------------
if __name__ == "__main__":
    np.random.seed(101)
    n_samples = 2500
    n_features = 5
    n_factors = 2

    # Generate an underlying core system driven by 2 hidden physical forces
    latent_f1 = np.random.normal(0, 1, n_samples)
    latent_f2 = np.random.normal(0, 1, n_samples)

    # Map hidden forces to 5 structural strain sensors with unique local white noise floors
    s1 = 0.85 * latent_f1 + 0.10 * latent_f2 + np.random.normal(0, 0.4, n_samples)
    s2 = 0.75 * latent_f1 + 0.15 * latent_f2 + np.random.normal(0, 0.5, n_samples)
    s3 = 0.12 * latent_f1 + 0.80 * latent_f2 + np.random.normal(0, 0.3, n_samples)
    s4 = 0.05 * latent_f1 + 0.90 * latent_f2 + np.random.normal(0, 0.2, n_samples)
    s5 = np.random.normal(5, 0.0, n_samples)  # Injected flatline zero-variance channel to test guardrail

    # Construct input dataframe
    raw_data = pd.DataFrame({
        'axial_strain_A': s1,
        'axial_strain_B': s2,
        'shear_stress_A': s3,
        'shear_stress_B': s4,
        'isolated_dead_sensor': s5
    })

    # Run the isolated diagnostic algorithm
    try:
        factor_scores, diagnostic_logs = run_factor_analysis_pipeline(raw_data, target_factors=n_factors)
    except ValueError as err:
        print(f"Pipeline Execution Aborted: {err}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set global plotting style for professional engineering reports
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams.update({
    'font.family': 'sans-serif',
    'font.size': 10,
    'axes.labelsize': 11,
    'axes.titlesize': 12,
    'xtick.labelsize': 9,
    'ytick.labelsize': 9,
    'figure.titlesize': 14,
    'patch.edgecolor': 'none'
})

# -------------------------------------------------------------------------
# 1. GENERATE REPRESENTATIVE EVALUATION DATASETS
# -------------------------------------------------------------------------
# Dimensions
m_pca = 6
k_range = np.arange(1, m_pca + 1)
sensors_pca = [f'Strain_{i+1}' for i in range(m_pca)]
pcs = [f'PC{j+1}' for j in range(m_pca)]

# Synthetic PCA Metrics
loadings_pca = np.array([
    [0.62, 0.51, 0.12, 0.05, 0.01, 0.01],
    [0.58, 0.54, -0.15, 0.02, 0.03, 0.01],
    [0.45, -0.48, 0.61, 0.10, 0.02, 0.02],
    [0.41, -0.52, -0.55, 0.15, 0.04, 0.01],
    [0.10, 0.12, 0.22, -0.78, 0.45, 0.11],
    [0.05, 0.08, 0.15, 0.52, 0.78, 0.22]
])
eigenvalues = np.array([2.85, 1.42, 0.88, 0.51, 0.24, 0.10])
marginal_variance = (eigenvalues / np.sum(eigenvalues)) * 100
cumulative_variance = np.cumsum(marginal_variance)
residual_variance = 100 - cumulative_variance

# Empirical expected values for distances across truncation sizes
expected_T2 = np.array([1.00, 2.00, 3.00, 4.00, 5.00, 6.00])
expected_Q = np.array([np.sum(eigenvalues[k:]) for k in range(1, m_pca + 1)])
expected_Q[-1] = 0.0  # Full subspace has zero residual

# Synthetic FA Metrics
m_fa = 5
factors_fa = ['Factor 1', 'Factor 2']
sensors_fa = ['Axial_A', 'Axial_B', 'Shear_A', 'Shear_B', 'Temp_Aux']

loadings_fa = np.array([
    [0.88, 0.12],
    [0.82, 0.15],
    [0.08, 0.84],
    [0.11, 0.79],
    [0.21, 0.18]
])
communality = np.sum(loadings_fa**2, axis=1) * 100
uniqueness = 100 - communality
factor_score_variances = np.array([1.12, 0.94])


# -------------------------------------------------------------------------
# 2. UI LAYOUT 1: PCA OPTIMIZATION SUITE (2 x 3 HORIZONTAL GRID)
# -------------------------------------------------------------------------
fig1, axs1 = plt.subplots(2, 3, figsize=(16, 10))
fig1.suptitle("PCA OPTIMIZATION & SUBSPACE DIAGNOSTIC SUITE", weight='bold', y=0.96)

# Panel 1,1: Heatmap of Feature Loadings Matrix |P_{j,r}|
sns.heatmap(np.abs(loadings_pca), annot=True, fmt=".2f", cmap="Blues", cbar=True,
            xticklabels=pcs, yticklabels=sensors_pca, ax=axs1[0, 0], cbar_kws={'label': 'Absolute Weight'})
axs1[0, 0].set_title("Feature Loadings Matrix $|P_{j,r}|$")
axs1[0, 0].set_xlabel("Principal Components")

# Panel 1,2: Absolute Eigenvalues Tracking Variance Magnitude
axs1[0, 1].bar(pcs, eigenvalues, color="#34495e", width=0.6, edgecolor='black', alpha=0.85)
axs1[0, 1].set_title("Absolute Eigenvalues ($\lambda_j$)")
axs1[0, 1].set_ylabel("Eigenvalue Magnitude")
axs1[0, 1].set_xlabel("Principal Components")

# Panel 1,3: Combined Bar + Line (Scree Plot)
axs1[0, 2].bar(pcs, marginal_variance, color="#2980b9", width=0.6, label="Marginal Variance %", alpha=0.8)
ax1_3_twin = axs1[0, 2].twinx()
ax1_3_twin.plot(pcs, cumulative_variance, color="#e74c3c", marker='o', linestyle='--', linewidth=2, label="Cumulative %")
ax1_3_twin.axhline(85, color='#27ae60', linestyle=':', alpha=0.7, label='85% Threshold')
axs1[0, 2].set_ylabel("Marginal Variance (%)")
ax1_3_twin.set_ylabel("Cumulative Variance (%)")
axs1[0, 2].set_title("Captured Structural Variance Profile")
# Merge legends cleanly
lines, labels = axs1[0, 2].get_legend_handles_labels()
lines2, labels2 = ax1_3_twin.get_legend_handles_labels()
axs1[0, 2].legend(lines + lines2, labels + labels2, loc='center right')
ax1_3_twin.grid(False)

# Panel 2,1: Residual Space Showing Excluded Information
axs1[1, 0].bar(pcs, residual_variance, color="#95a5a6", width=0.6, edgecolor='black', alpha=0.8)
axs1[1, 0].set_title("Residual Space (Unexplained Variance %)")
axs1[1, 0].set_ylabel("Discarded Energy Proportion (%)")
axs1[1, 0].set_xlabel("Subspace Cutoff Boundary ($k$)")
axs1[1, 0].set_ylim(0, 105)

# Panel 2,2: Sample Mean Hotelling's T^2 vs Subspace Size k
axs1[1, 1].plot(k_range, expected_T2, color="#16a085", marker='s', linewidth=2, markersize=6)
axs1[1, 1].set_title("Empirical Mean Hotelling's $T^2$ vs. $k$")
axs1[1, 1].set_ylabel("Expectation $\mathbb{E}[T^2]$")
axs1[1, 1].set_xlabel("Subspace Dimension ($k$)")
axs1[1, 1].set_xticks(k_range)

# Panel 2,3: Sample Mean Q Statistic vs Truncation Cutoff k
axs1[1, 2].plot(k_range, expected_Q, color="#d35400", marker='d', linewidth=2, markersize=6)
axs1[1, 2].set_title("Empirical Mean $Q$ Statistic (SPE) vs. $k$")
axs1[1, 2].set_ylabel("Expectation $\mathbb{E}[Q]$")
axs1[1, 1].set_xlabel("Subspace Dimension ($k$)")
axs1[1, 2].set_xlabel("Truncation Cutoff Boundary ($k$)")
axs1[1, 2].set_xticks(k_range)

plt.tight_layout(rect=[0, 0, 1, 0.93])
plt.show()


# -------------------------------------------------------------------------
# 3. UI LAYOUT 2: FA LATENT SUBSPACE SUITE (2 x 2 GRID MATRIX)
# -------------------------------------------------------------------------
fig2, axs2 = plt.subplots(2, 2, figsize=(14, 9))
fig2.suptitle("FACTOR ANALYSIS LATENT SUBSPACE DIAGNOSTIC SUITE", weight='bold', y=0.96)

# Panel 1,1: Heatmap of Structural Loadings Matrix |\lambda_{j,r}|
sns.heatmap(np.abs(loadings_fa), annot=True, fmt=".2f", cmap="Purples", cbar=True,
            xticklabels=factors_fa, yticklabels=sensors_fa, ax=axs2[0, 0], cbar_kws={'label': 'Absolute Loading'})
axs2[0, 0].set_title("Structural Loadings Matrix $|\boldsymbol{\Lambda}_{\text{rotated}}|$")
axs2[0, 0].set_xlabel("Rotated Common Factors")

# Panel 1,2: Stacked Horizontal Bar Chart of Variance Allocation
y_pos = np.arange(len(sensors_fa))
axs2[0, 1].barh(y_pos, communality, color="#2ecc71", edgecolor='none', height=0.5, label="Shared Communality ($h^2$)")
axs2[0, 1].barh(y_pos, uniqueness, left=communality, color="#e74c3c", edgecolor='none', height=0.5, label="Channel Uniqueness ($\psi^2$)")
axs2[0, 1].set_yticks(y_pos)
axs2[0, 1].set_yticklabels(sensors_fa)
axs2[0, 1].set_xlabel("Variance Percentage Distribution (%)")
axs2[0, 1].set_title("Variance Allocation Breakdown ($h^2$ vs. $\psi^2$)")
axs2[0, 1].set_xlim(0, 100)
axs2[0, 1].legend(loc="lower left")

# Panel 2,1: Sensor Uniqueness Noise Floor Profile across Monitored Channels
axs2[1, 0].plot(sensors_fa, uniqueness, color="#7f8c8d", marker='o', linestyle='-', linewidth=2, markersize=7)
# Highlight potential instrument fault zone (e.g., Temp_Aux has very high uniqueness)
axs2[1, 0].fill_between(sensors_fa, uniqueness, color="#f1c40f", alpha=0.15)
axs2[1, 0].set_title("Sensor Uniqueness Noise Floor Profile ($\psi^2$)")
axs2[1, 0].set_ylabel("Uniqueness Level (%)")
axs2[1, 0].set_ylim(0, 105)
axs2[1, 0].set_xlabel("Monitored Sensor Channels")
for i, val in enumerate(uniqueness):
    axs2[1, 0].text(i, val + 3, f"{val:.1f}%", ha='center', va='bottom', fontsize=9, weight='semibold')

# Panel 2,2: Latent Factor Scores Empirical Variance
axs2[1, 1].bar(factors_fa, factor_score_variances, color="#8e44ad", width=0.4, edgecolor='black', alpha=0.85)
axs2[1, 1].axhline(1.0, color='black', linestyle=':', alpha=0.5, label='Pre-Rotation Baseline')
axs2[1, 1].set_title("Latent Factor Scores Post-Rotation Empirical Variance")
axs2[1, 1].set_ylabel("Empirical Score Variance ($\operatorname{Var}(F_l)$)")
axs2[1, 1].set_ylim(0, np.max(factor_score_variances) * 1.3)
axs2[1, 1].legend(loc="upper right")

plt.tight_layout(rect=[0, 0, 1, 0.93])
plt.show()
```

Part B: Analysis

Question 1: The Total Variance Illusion (PCA Subplots 1 & 2 vs. FA Subplots 1 & 2)

## 1. Physical Interpretation of Sensor 4's Uniqueness ($\varphi^2 \approx 98\%$)

In Factor Analysis, a sensor's total variance is decomposed into a shared component (Communality, $h_j^2$) and an isolated component (Uniqueness, $\varphi_j^2$).

A uniqueness value exceeding $98\%$ explicitly demonstrates that **almost none of the variance captured by Sensor 4 is driven by the underlying physical forces ($f_1$ and $f_2$)** regulating the rest of the asset.

* **Physical Reality:** Sensor 4 has effectively decoupled from the structural and operational dynamics of the plant.
* **Diagnostic Meaning:** Because its variance is entirely self-contained and unshared, it represents a severe localized instrumentation failure—such as sensor degradation, an electrical ground loop, or severe electromagnetic interference—rather than a systemic structural shift.

---

## 2. How the Noise Floor "Tricks" the PCA Engine

Principal Component Analysis (PCA) works by maximizing *total global variance*, making no mathematical distinction between meaningful shared variance (covariance) and meaningless localized variance (noise variance).

$$\operatorname{Var}(X_{\text{global}}) = \text{Shared Variance} + \text{Localized Noise}$$

Because Sensor 4 is contaminated with an exceptionally high noise scale ($\sigma^2 \approx 2.0$), it introduces a massive spike of raw energy into the system.

Question 2: Decoupling Structural Loading via Rotation (FA Subplot 1 vs. PCA Eigenvectors)

## 1. Mathematical Simplification: Moving to a "Simple Structure"

Traditional Principal Component Analysis (PCA) operates under a strict maximization bottleneck: it forces the first principal axis to align with the direction of maximum global variance, the second to capture the maximum remaining orthogonal variance, and so on. This mathematical hierarchy ($\lambda_1 > \lambda_2 > \dots > \lambda_m$) means that the resulting eigenvectors are almost always **linear combinations of all available sensors**, leading to highly cross-contaminated and complex loading profiles.

Factor Analysis with **Varimax rotation** breaks this rigid hierarchy. Varimax is an orthogonal rotation method that maximizes the *variance of the squared loadings* within each factor column:

$$V = \frac{1}{m} \sum_{l=1}^k \left[ m \sum_{j=1}^m (\lambda_{jl}^2)^2 - \left( \sum_{j=1}^m \lambda_{jl}^2 \right)^2 \right]$$



By maximizing this metric, the algorithm drives individual factor loadings toward the extremes of the scale—either pushing them close to $\pm 1$ or forcing them down near $0$. This mathematical shift relaxes the global variance constraint in favor of a **Simple Structure**. Instead of forcing every component to sweep up mixed signals, it isolates and groups variables together, establishing distinct, non-overlapping blocks of heavily loaded sensors.

---

## 2. Control Room Troubleshooting: FA Heatmap vs. Raw PCA

From a plant operator's standpoint, translating abstract statistical charts into immediate physical maintenance actions is critical. The structural clarity of a rotated Factor Analysis heatmap provides a major advantage over a raw PCA matrix during a live asset failure:

### The Raw PCA Matrix (Complex Mixture)
In an unrotated PCA eigenvector matrix, a structural shift in the plant will project across multiple principal components simultaneously. Because every component contains mixed contributions from all sensors, an operator looking at an anomaly in `PC 1` will see fractional weights from `Sensor_1`, `Sensor_2`, and `Sensor_3` mixed together. It becomes mathematically challenging to untangle which physical component is actually failing, creating an ambiguous diagnostic signal that delays targeted maintenance.

### The Rotated FA Heatmap (Physical Isolation)
Because Varimax rotation isolates distinct blocks, it maps the underlying mathematical factors directly to **isolated physical subsystems** or mechanical degrees of freedom within the plant:

| Latent Dimension | Heavy Sensor Loading | Physical Component Mapping |
| :--- | :--- | :--- |
| **Factor 1** | `Sensor_1` & `Sensor_2` | **Primary Structural Truss Bending Mode** |
| **Factor 2** | `Sensor_3` | **Secondary Torsional Drive Operational Mode** |



### Live Failure Diagnostic Scenario
If a physical weld fails or a structural fatigue crack propagates along the primary truss structure:
* **The PCA Dashboard** will show a generalized, muddy spike across `PC 1` and `PC 2`, indicating that the entire multi-sensor covariance engine is drifting, without pointing to a clear root cause.
* **The FA Dashboard** will pinpoint the issue instantly. The operator will observe a sharp, isolated anomaly focused exclusively on **Factor 1** (`Sensor_1` and `Sensor_2`), while **Factor 2** (`Sensor_3`) remains completely green and nominal.

This clean isolation tells the control room team exactly where to look, reducing diagnostic search times from hours to seconds and preventing unnecessary plant shutdowns.

Question 3: Determining Subspace Truncation ( k ) using  T2  and  Q  Profiles

## 1. Dynamic Behavior of the Mean $Q$ Statistic Curve

The Mean $Q$ Statistic (Squared Prediction Error) measures the squared Euclidean distance from an incoming data snapshot to its projection on the $k$-dimensional principal subspace, effectively tracking the energy left in the residual workspace.

As the truncation boundary transitions across cutoff sizes, the curve exhibits two distinct behavioral phases:

* **Transition from $k=1$ to $k=2$:** The curve undergoes a **massive, steep downward drop**. This vertical plunge indicates that adding the second principal component captures a massive block of structured variance, stripping it out of the residual space and drastically reducing the projection error.
* **Transition from $k=2$ to $k=3$ (and beyond):** The curve suddenly flattens out, forming a distinct **horizontal "flat elbow"**. The transition from $k=2$ to $k=3$ yields a negligible decrease in the mean $Q$ value, showing that any components added past this point account for very little remaining variance.

---

## 2. Identifying the Asset's Latent Physical Dimensionality ($k=2$)

The structural transition point—the sharp drop followed by the flat elbow—serves as a clear mathematical indicator of the system's true latent dimensionality for three reasons:

* **Signal vs. Noise Partitioning:** The massive drop before $k=2$ proves that the first two components contain **true structural signal** (the common physical driving forces, $f_1$ and $f_2$, which represent the primary loading and secondary operational modes of the asset).
* **Exhaustion of Macro-Variations:** The flat floor after $k=2$ demonstrates that the underlying physical processes driving the multi-sensor correlation network have been completely exhausted.
* **The Noise Baseline:** The remaining residual energy forming the elbow is composed entirely of unshared, isotropic background white noise. Because random noise cannot be compressed into organized dimensions, adding components past $k=2$ yields no meaningful reduction in projection error, confirming that **$k=2$** is the optimal boundary for the asset model.

---

## 3. The Consequences of Over-Fitting ($k=3$)

If you incorrectly select a truncation cutoff of $k=3$, you are over-fitting the model space and compromising the integrity of the diagnostic system:



* **Forcing Noise into the Signal Subspace:** By expanding the active tracking plane to $k=3$, you force the third principal axis to capture the random, high-frequency background noise floor. This incorporates unmodeled instrumentation noise or environmental hiss into your "clean" structural subspace.
* **Loss of Subspace Sensitivity:** A clean principal subspace should only change when macro-level physical forces drift. By polluting it with random noise variations, the baseline coordinates become unstable and diluted.
* **Diagnostic Blindness:** When a true structural failure occurs, its early-stage variance can get misclassified or absorbed by this extra noise-polluted component. This artificially dampens the response of the residual $Q$ statistic, creating a dangerous **blind spot (Type II error)** that masks real physical degradation.

Question 4: Operational Trade-offs in System Health Monitoring

## Operational Strategy Comparison: PCA vs. Factor Analysis

Deploying a real-time predictive maintenance pipeline for a fleet of identical industrial assets requires balancing macro-level structural tracking with localized instrument robustness. Below is a detailed breakdown of the operational trade-offs between the two approaches.

### 1. The PCA Strategy ($T^2$ and $Q$ Statistics)
The PCA strategy separates the multi-sensor observation space into two orthogonal sub-environments: an in-subspace energy plane (Hotelling's $T^2$) and an out-of-subspace residual plane ($Q$ statistic).

* **Operational Strengths:**
    * **Comprehensive Macro Tracking:** Excellent at capturing unexpected, unmodeled global failures. Since the residual $Q$ statistic monitors everything outside the primary subspace, it functions as a universal tripwire for any change that alters the baseline correlation structure.
    * **Minimal Domain Assumptions:** PCA does not require an underlying structural model; it simply optimizes for total global variance, making it highly efficient to scale across a large fleet.
* **Operational Weaknesses:**
    * **Diagnostic Ambiguity:** When an alert is triggered, a high $T^2$ or $Q$ value only indicates *that* a variance shift has occurred. Determining *which* component or sensor failed requires secondary, computationally intensive root-cause analysis (such as contribution plots).

---

### 2. The FA Strategy (Rotated Latent Factor Scores)
The Factor Analysis strategy uses a generative model to isolate a specific number of common physical forces while explicitly accounting for localized sensor variance.

* **Operational Strengths:**
    * **Immediate Root-Cause Diagnostics:** Thanks to the Varimax rotation, each common factor tracks an isolated physical subsystem or operational mode. If a failure occurs, the specific factor score spikes instantly, telling operators exactly which mechanical component requires maintenance without further troubleshooting.
    * **Structured Noise Isolation:** The framework separates global structural trends from localized background noise, creating a highly stable tracking environment for the primary asset dynamics.
* **Operational Weaknesses:**
    * **Blindness to Novel Fault Vectors:** FA is bounded by its pre-defined latent factor structure. If the asset encounters a unique, unmodeled failure mode that falls completely outside the scope of the common factors, the system may fail to capture it, running the risk of a missed detection ($\beta$).

---

## Robustness Mapping: Handling Sensor Faults

When a single sensor loses calibration or experiences an electrical short, **The Factor Analysis (FA) Strategy is significantly more robust** than the PCA strategy.

### Mechanical and Statistical Justification

The core advantage of the FA strategy lies in the **Sensor Uniqueness Noise Floor ($\varphi^2$)** parameter matrix, which is entirely absent from the PCA architecture:

$$\mathbf{R} = \boldsymbol{\Lambda}\boldsymbol{\Lambda}^T + \boldsymbol{\Psi}$$



1.  **Isolation via the Diagonal Matrix ($\boldsymbol{\Psi}$):** The Factor Analysis engine explicitly dedicates a diagonal workspace, $\boldsymbol{\Psi} = \operatorname{diag}(\varphi_1^2, \dots, \varphi_m^2)$, to capture the independent, unshared variance of each individual sensor channel.
2.  **Absorbing the Fault Energy:** If a sensor suffers an electrical short or calibration drift, it injects a massive spike of localized, unshared variance into the system. The FA engine dynamically isolates this failure by expanding that specific channel's **Uniqueness value ($\varphi^2$) toward 100%**. Because this noise is funneled into the diagonal matrix, it is prevented from polluting the shared loading matrix $\boldsymbol{\Lambda}$.
3.  **Stability of the Core Fleet Monitors:** Since the loading matrix remains untainted, the rotated Latent Factor Scores continue to accurately track the true physical driving forces of the plant. The asset's structural health monitors remain completely stable, unaffected by the isolated instrument fault.

### The Vulnerability of PCA under the Same Fault
In contrast, the PCA strategy collapses under a severe localized sensor fault. Because PCA maximizes total global variance without distinguishing between types of variance, it cannot isolate unshared noise.

The massive energy spike from the shorted sensor will distort the covariance matrix, dragging the dominant principal components out of alignment with the true physical axes. This creates an immediate system-wide failure: Hotelling's $T^2$ and the residual $Q$ statistic will both spike catastrophically, triggering a **false alarm ($\alpha$)** and forcing an unnecessary plant shutdown due to a single bad instrument.
