# Question 1: Statistical Mechanics of the Quantum Harmonic Oscillator

## Given

A quantum harmonic oscillator has energy eigenvalues

$
E_j=\hbar\omega\left(j+\frac12\right),\qquad j=0,1,2,\ldots
$

where

* $\hbar$ = Reduced Planck's constant
* $\omega$ = Angular frequency
* $k_B$ = Boltzmann constant
* $T$ = Absolute temperature

The inverse temperature is

$
\beta=\frac{1}{k_BT}
$


---

# 1. Partition Function

The canonical partition function is

$
Z(T)=\sum_{j=0}^{\infty}e^{-E_j/(k_BT)}
$

Substitute

$
E_j=\hbar\omega\left(j+\frac12\right)
$

Therefore,

$
Z=\sum_{j=0}^{\infty}
e^{-\frac{\hbar\omega(j+\frac12)}{k_BT}}
$

Separate the exponential terms:

$
Z=
e^{-\frac{\hbar\omega}{2k_BT}}
\sum_{j=0}^{\infty}
\left(e^{-\frac{\hbar\omega}{k_BT}}\right)^j
$

The series

$
1+r+r^2+\cdots
$

is a geometric series with

$
r=e^{-\frac{\hbar\omega}{k_BT}}
$

Since

$
0<r<1,
$

the series converges to

$
\frac1{1-r}
$

Hence,

$
\boxed{
Z=
\frac{
e^{-\frac{\hbar\omega}{2k_BT}}
}{
1-e^{-\frac{\hbar\omega}{k_BT}}
}
}
$

**Answer:** The partition function is

$
\boxed{
Z(T)=
\frac{
e^{-\frac{\hbar\omega}{2k_BT}}
}{
1-e^{-\frac{\hbar\omega}{k_BT}}
}
}
$

---

# 2. Helmholtz Free Energy

The Helmholtz free energy is

$
F=-k_BT\ln Z
$


Simplifying,

$
\boxed{
F=
\frac{\hbar\omega}{2}
+
k_BT
\ln
\left(
1-e^{-\frac{\hbar\omega}{k_BT}}
\right)
}
$

---

# 3. Internal Energy

The internal energy is

$
U=-\frac{\partial\ln Z}{\partial\beta}
$

where

$
\beta=\frac1{k_BT}
$



Thus,

$
\boxed{
U=
\hbar\omega
\left(
\frac12
+
\frac1
{e^{\frac{\hbar\omega}{k_BT}}-1}
\right)
}
$

---

## 4. Entropy

The entropy is defined as

\$
S=-\left(\frac{\partial F}{\partial T}\right)_V
\$

Using the free energy expression, we obtain

$
\boxed{
S
=
-k_B \ln\left(1-e^{-\frac{\hbar\omega}{k_B T}}\right)
+
\frac{\hbar\omega}{T}
\frac{1}{e^{\frac{\hbar\omega}{k_B T}}-1}
}
$

The entropy also satisfies the thermodynamic identity

\$
\boxed{
U = F + TS
}
\$

which confirms that the derivation is correct.

---

## 5. Heat Capacity

The heat capacity at constant volume is defined as

\$
C_V=\left(\frac{\partial U}{\partial T}\right)_V
\$

Differentiating the internal energy expression with respect to temperature gives

$
\boxed{
C_V
=
k_B
\left(\frac{\hbar\omega}{k_B T}\right)^2
\frac{\exp\left(\frac{\hbar\omega}{k_B T}\right)}
{\left[\exp\left(\frac{\hbar\omega}{k_B T}\right)-1\right]^2}
}
$

Using the hyperbolic sine identity

$
\sinh(x)=\frac{e^x-e^{-x}}{2},
$

the heat capacity can also be written as

$
\boxed{
C_V
=
k_B
\left(\frac{\hbar\omega}{2k_B T}\right)^2
\frac{1}
{\sinh^2\left(\frac{\hbar\omega}{2k_B T}\right)}
}
$


---

# 6. High-Temperature Limit

When

$
k_BT\gg\hbar\omega,
$

the exponential term may be approximated by

$
e^{\frac{\hbar\omega}{k_BT}}
\approx
1+\frac{\hbar\omega}{k_BT}
$

Substituting this approximation into the heat-capacity equation gives

$
\boxed{
C_V\approx k_B
}
$






## Question 2: The Maximum Entropy Distribution under Mean and Variance Constraints

### Given

Let (X) be a continuous random variable with

* Mean:

$
E[X]=\mu
$

* Variance:

$
E[(X-\mu)^2]=\sigma^2
$

Show that the probability distribution with the maximum entropy is the Gaussian (Normal) distribution.

---

# Solution

## Step 1: Shannon-Jaynes Entropy

For a continuous random variable, the entropy is

$
H[p]=-\int_{-\infty}^{\infty}p(x)\ln p(x),dx
$

Our objective is

$
\boxed{\text{Maximize }H[p]}
$

subject to the given constraints.

---

# Step 2: Apply the Constraints

The probability density function must satisfy three conditions.

### (a) Normalization

The total probability must equal one.

$
\boxed{\int_{-\infty}^{\infty}p(x),dx=1}
\$

---

### (b) Mean Constraint

The average value is

$
\boxed{\int_{-\infty}^{\infty}xp(x),dx=\mu}
$

---

### (c) Variance Constraint

The variance is

$
\boxed{\int_{-\infty}^{\infty}(x-\mu)^2p(x),dx=\sigma^2}
$

---

# Step 3: Construct the Lagrangian

Introduce three Lagrange multipliers:

* $\gamma$ (normalization)
* $\lambda_1$ (mean)
* $\lambda_2$ (variance)

The Lagrangian is

$
\begin{aligned}
L=&-\int p(x)\ln p(x),dx\
&-\gamma\left(\int p(x),dx-1\right)\
&-\lambda_1\left(\int xp(x),dx-\mu\right)\
&-\lambda_2\left(\int(x-\mu)^2p(x),dx-\sigma^2\right)
\end{aligned}
$

---

## Step 4: Differentiate

Take the functional derivative with respect to \(p(x)\):

\$
\frac{\delta L}{\delta p(x)} = 0
\$

This gives

$
-\ln p(x) - 1 - \gamma - \lambda_1 x - \lambda_2 (x-\mu)^2 = 0.
$

Hence,

$
\ln p(x)
=
-1 - \gamma - \lambda_1 x - \lambda_2 (x-\mu)^2.
$

Exponentiating both sides,

$
p(x)
=
e^{-1-\gamma}
\,e^{-\lambda_1 x}
\,e^{-\lambda_2 (x-\mu)^2}.
$

Let

\$
Z=e^{1+\gamma}.
\$

Then,

$
\boxed{
p(x)
=
\frac{1}{Z}
\exp\!\left[-\lambda_1 x-\lambda_2 (x-\mu)^2\right]
}
$

---

# Step 5: Determine the Multipliers

Using the mean constraint,

$
E[X]=\mu
$

the linear term disappears.

Therefore,

$
\boxed{\lambda_1=0}
$

Using the variance constraint,

$
\boxed{\lambda_2=\frac1{2\sigma^2}}
$

---

# Step 6: Determine the Normalization Constant

The normalization condition requires

$
\int p(x),dx=1
$

Evaluating the Gaussian integral gives

$
\boxed{
Z=\sqrt{2\pi\sigma^2}
}
$

---

## Step 7: Final Probability Density Function

Substitute the values of the Lagrange multipliers into the expression for \(p(x)\). The resulting probability density function is

$
\boxed{
p(x)
=
\frac{1}{\sqrt{2\pi\sigma^2}}
\exp\left(
-\frac{(x-\mu)^2}{2\sigma^2}
\right)
}
$






## Question 3: Multiclass Logistic Regression – Automatic Classification of Emails

### Given

An e-commerce platform automatically classifies customer emails into one of several categories using a Maximum Entropy (MaxEnt) / Multiclass Logistic Regression model.

The input feature vector is

$
x=(x_1,x_2,x_3)^T
$

where

* (x_1) = Financial keywords detected
* (x_2) = Account access keywords detected
* (x_3) = Technical/Bug keywords detected

The five feature scenarios and their probabilities are:

| Scenario | Feature Vector | Probability |
| -------- | -------------- | ----------- |
| A        | (1,0,0)        | 0.40        |
| B        | (0,1,0)        | 0.25        |
| C        | (0,0,1)        | 0.15        |
| D        | (0,1,1)        | 0.15        |
| E        | (1,1,0)        | 0.05        |

The observed constraints are

$
E[x_1I(Y=1)]=0.05
$

$
E[x_2I(Y=1)]=0.15
$

$
E[x_3I(Y=1)]=0.16
$

These constraints represent the historical routing behavior of emails.

---

# Step 1: Maximum Entropy Model

For binary classification,

$
P(Y=1|x)=
\frac{e^{w\cdot x}}
{1+e^{w\cdot x}}
$

where

$
w=(w_1,w_2,w_3)
$

is the weight vector.

Similarly,

$
P(Y=0|x)=
\frac1{1+e^{w\cdot x}}
$

This is called the **sigmoid (logistic) function**.

---

# Step 2: Calculate the Linear Score

For each scenario,

$
z=w\cdot x
$

### Scenario A

$
x=(1,0,0)
$

Therefore,

$
z_A=w_1
$

Hence,

$
P_A=
\frac{1}{1+e^{-w_1}}
$

---

### Scenario B

$
x=(0,1,0)
$

$
z_B=w_2
$

Therefore,

$
P_B=
\frac1{1+e^{-w_2}}
$

---

### Scenario C

$
x=(0,0,1)
$

$
z_C=w_3
$

Thus,

$
P_C=
\frac1{1+e^{-w_3}}
$

---

### Scenario D

$
x=(0,1,1)
$

$
z_D=w_2+w_3
$

Therefore,

$
P_D=
\frac1{1+e^{-(w_2+w_3)}}
$

---

### Scenario E

$
x=(1,1,0)
$

$
z_E=w_1+w_2
$

Hence,

$
P_E=
\frac1{1+e^{-(w_1+w_2)}}
$

---

# Step 3: Apply the Historical Constraints

## Constraint 1

Financial emails

[
E[x_1I(Y=1)]
============

0.05
]

Only Scenarios A and E contribute because they contain (x_1=1).

Therefore,

[
0.40P_A+0.05P_E=0.05
]

---

## Constraint 2

Account access emails

Scenarios B, D and E contribute.

Hence,

[
0.25P_B
+
0.15P_D
+
0.05P_E
=======

0.15
]

---

## Constraint 3

Technical emails

Only Scenarios C and D contribute.

Therefore,

[
0.15P_C
+
0.15P_D
=======

0.16
]

These three equations determine the unknown weights $w_1,w_2,w_3$. In practice, they are solved numerically using optimization algorithms such as Gradient Descent or L-BFGS.

---

## Step 4: Classification Rule

After solving for the weights, calculate

$
z=\mathbf{w}\cdot\mathbf{x}
$

Then compute the probability using the logistic (sigmoid) function:

$
P(Y=1\mid x)
=
\frac{1}{1+e^{-z}}
$

The classification decision is based on the following rule:

- If

  $
  P(Y=1\mid x)\ge 0.5,
  $

  classify the email as **Technical Support**.

- Otherwise,

  $
  P(Y=1\mid x)<0.5,
  $

  classify the email as **Finance/Other Support**.

---

# Step 5: Interpretation of the Constraints

* $E[x_1I(Y=1)]=0.05$: Emails with financial keywords are rarely routed to Technical Support.
* $E[x_2I(Y=1)]=0.15$: Account-access emails are moderately likely to be routed to Technical Support.
* $E[x_3I(Y=1)]=0.16$: Emails mentioning bugs or malfunctions have the highest probability of being routed to Technical Support.

---

# Advantages of the Maximum Entropy Model

* Produces the least-biased probability distribution consistent with the observed data.
* Works well for text classification problems.
* Estimates class probabilities rather than only predicting labels.
* Supports multiple input features simultaneously.
* Uses historical constraints without making unnecessary assumptions.



# Question 4: Asset Pricing and Portfolio Optimization Using the Maximum Entropy Principle

## Given

Asset X is currently trading at **$100**. We want to determine the **risk-neutral probability distribution** of the future asset price (S_T) after one month using the Maximum Entropy (MaxEnt) principle.

The available market information (constraints) is:

* Expected future price:

$
E[S_T]=102
$

* Expected squared price:

$
E[S_T^2]=10600
$

* Expected logarithmic price:

$
E[\ln(S_T)]=4.60
$

Our objective is to determine the probability density function (p(S_T)) that satisfies these constraints while maximizing entropy.

---

# Solution

## Step 1: Define the Entropy Function

For a continuous random variable, the **Shannon entropy** is defined as

$
H[p]
=
-\int_{0}^{\infty} p(S_T)\,\ln p(S_T)\,dS_T.
$

The objective is to maximize the entropy

$
\boxed{\text{Maximize } H[p]}
$

subject to the given constraints.

---

# Step 2: Apply the Constraints

The probability density must satisfy the following conditions.

### (a) Normalization

$
\boxed{
\int_0^\infty p(S_T),dS_T=1
}
$

---

### (b) Expected Future Price

$
\boxed{
\int_0^\infty S_T,p(S_T),dS_T=102
}
$

---

### (c) Expected Squared Price

$
\boxed{
\int_0^\infty S_T^2,p(S_T),dS_T=10600
}
$

---

### (d) Logarithmic Expectation

$
\boxed{
\int_0^\infty \ln(S_T),p(S_T),dS_T=4.60
}
$

These four equations are the market constraints that the probability distribution must satisfy.

---

# Step 3: Construct the Maximum Entropy Lagrangian

Introduce the Lagrange multipliers

* $\gamma$ for normalization
* $\lambda_1$ for the mean
* $\lambda_2$ for the second moment
* $\lambda_3$ for the logarithmic expectation

The Lagrangian is


\begin{aligned}
L
=&
-\int p\ln p,dS_T\
&
-\gamma\left(\int p,dS_T-1\right)\
&
-\lambda_1\left(\int S_Tp,dS_T-102\right)\
&
-\lambda_2\left(\int S_T^2p,dS_T-10600\right)\
&
-\lambda_3\left(\int\ln(S_T)p,dS_T-4.60\right)
\end{aligned}


---

## Step 4: Differentiate

Take the functional derivative with respect to \(p(S_T)\):

\$
\frac{\delta L}{\delta p(S_T)}=0.
\$

This gives

\$
-\ln p(S_T)-1-\gamma-\lambda_1S_T-\lambda_2S_T^2-\lambda_3\ln(S_T)=0.
\$

Rearranging,

$
\ln p(S_T)
=
-1-\gamma-\lambda_1S_T-\lambda_2S_T^2-\lambda_3\ln(S_T).
$

Exponentiating both sides,

$
p(S_T)
=
e^{-1-\gamma}
\,S_T^{-\lambda_3}
\,e^{-\lambda_1S_T}
\,e^{-\lambda_2S_T^2}.
$

---

## Step 5: Define the Partition Function

Let

\$
Z=e^{\,1+\gamma}.
\$

Then,

$
\boxed{
p(S_T)
=
\frac{1}{Z}
S_T^{-\lambda_3}
\exp\!\left(-\lambda_1S_T-\lambda_2S_T^2\right)
}
$

where the partition function is

$
Z
=
\int_{0}^{\infty}
S_T^{-\lambda_3}
\exp\!\left(-\lambda_1S_T-\lambda_2S_T^2\right)
\,dS_T.
$

This partition function ensures that the total probability equals one,

\$
\int_{0}^{\infty} p(S_T)\,dS_T = 1.
\$

---

# Step 6: Interpretation

Each market constraint has a physical meaning.

### Expected Price

Keeps the average future asset price equal to the forward price.

### Second Moment

Controls the market volatility (risk).

### Logarithmic Constraint

Captures downside crash risk and market skewness.

Unlike a normal distribution, this model allows asymmetric probability distributions that better represent real financial markets.

---

# Step 7: Numerical Solution

The parameters

$
\lambda_1,\lambda_2,\lambda_3
$

cannot usually be obtained analytically.

Instead, they are estimated numerically using optimization methods such as:

* Newton-Raphson
* Gradient Descent
* L-BFGS

The optimization continues until all market constraints are satisfied simultaneously.

---

# Advantages of Maximum Entropy in Finance

* Produces the least-biased probability distribution.
* Uses only available market information.
* Captures skewness and fat tails.
* Improves derivative pricing.
* Better reflects real market behavior than assuming a normal distribution.
* Widely used in quantitative finance and portfolio optimization.
